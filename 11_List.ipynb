**append** vs **extend** vs **concat**

- apppend will add the the elements at last, that elements can be any type

         - list1=[1,2,3]

         - list2=['A','B']

         - list1.append(list2)=[1,2,3,['A','B']]

  

- Extend will update the list by adding new elements,the result also save in a same list

         - list1=[1,2,3]

         - list2=['A','B']

         - list1.extend(list2)=[1,2,3,'A','B']

  

- concate will add two list,but the result will not update in a same list

         - list1=[1,2,3]

         - list2=['A','B']

         - list1+list2=[1,2,3,'A','B']
