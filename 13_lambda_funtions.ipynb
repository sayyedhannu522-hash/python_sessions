### lambda funtions

```python
# Create  basic function
```

**one argument**

```python
def add(x):
    return(x+10)

add(20)
```

```python
def add(x):
    return(x+10)

add(20)

# how many arguments are present?  :x
# what you are returning? : x+10

# format : lambda <argument_name>:<output>
```

```python
add=lambda x:x+10
add(20)
```

```python
def square(x):
    return(x*x)
square(10)
```

```python
square=lambda x:x*x
square(10)
```

```python
def cube(x):
    return(x*x*x)
cube(30)
```

```python
cube=lambda x:x*x*x
cube(30)
```

**two argumnets**

```python
def add(a,b):
    return(a+b)

add(20,30)

# how many arguments are present?  :a,b
# what you are returning? : a+b

# format : lambda <argument_name>:<output>
```

```python
add=lambda a,b:a+b
add(20,30)
```

```python
# implement average of 3 numbers using lambda
avg=lambda a,b,c:(a+b+c)/3
avg(5,5,5)
```

```python
# implement average make c as default parameter
avgr=lambda a,b,c=30:round((a+b+c)/3,2)
avgr(20,30)
```

**if-else**

```python
# Create a funtion for finding greater number between two numbers
def greater(n1,n2):
    if n1>n2:
        return(n1)
    else:
        return(n2)
greater (500,200)
```

```python
l1=[]
def greater(n1,n2):
    if n1>n2:
        l1.append(n1)
    else:
        l1.append(n2)
greater (500,200)
```

```python
l1
```

```python
def greater(n1,n2):
    if n1>n2:
        return(n1)
    else:
        return(n2)
greater (500,200)
# format: lambda <arg1>,<arg2>: <if_output>  <if_con>  else <else_op>
```

```python
greater=lambda n1,n2:(n1 if n1>n2 else n2)
greater(500,200)
```

```python
list1=[1,2,3,4,5]
# output:[1,4,9,16,25]
lambda i:i*i,list1
map(lambda i:i*i,list1)
list(map(lambda i:i*i,list1))
```

```python
list1=[1,2,3]
list2=[11,22,33]
# [12,24,36]
tuple(map(lambda i,j:i+j,list1,list2))
```

```python
list1=['h#d','mum#bai','chennai']
# output:['h#d','mum#bai']
list(map(lambda i: '#' in i,list1))

# condition mapping to list of items
```
