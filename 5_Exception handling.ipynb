### Exception  block

```python
#### try -except
- there are two blocks 
- one try block
- another is exception block
- your code will be there in try block
- error capture will happen in except block
- indentation : space
- red : space/indentation not maintained properly
- indentation has minimum  4 characters 
- : is there hence indentation  will be there
```

```python
try:
   num1=10
   num2=20
   num3=30
   add=num1+num2+num3
   print(add)
    
except:
    print("error is there , check code")
```

```python
# wap ask the user enter a number and find the square of the number
try:
    num=eval(input("enter number"))
    square=num*num
    print("the square of the {} is {}".format(num,square))
except Exception as e:
    print(e)   # capture error
    print('hello') # print 
```

- if your is with out error , try block will execute
- if your code has errors,it will redirect to except
- what ever you written inside except block that will run
- try and exception is together
- once try and exceptions code are complete , it try look any other lines of code are there.
