```python
# wap ask the user enter percentage of marks
# if percentage >90 print A
# if percentage between 70 to 90 print B
# if percentage between 50 to 70 print C
# if-elif-elif-else
num=eval(input("enter a number"))
if num>90:
    print("A")
elif 70<num<90:
    print("B")
elif 50<num<70:
    print("C")
else:
    print("ZERO")
```

```python
# wap ask the user enter which operation you need to perform 
#                              operation are : addition/mul/sub/div
# ask the user enter a number1
# ask the user enter a number2
# if the operation equal to addition
#                  print("we are doing addition operation")
#                  print("the addition of num1 and num2 is")
# if the operation equal to subtraction
#                  print("we are doing subtraction operation")
#                  print("the subtraction of num1 and num2 is")
# if the operation equal to multipulication
#                  print("we are doing multipulication operation")
#                  print("the multipulication of num1 and num2 is")
# if the operation equal to division
#                  print("we are doing division operation")
#                  print("the division of num1 and num2 is") 
# else:
#     print("provide suitable operation")
operation=(input("enter which want to operation"))
num1=eval(input("enter a number:"))
num2=eval(input("enter another number:"))
if operation=='addition':
    print("we are doing addition")
    print("the addition of num1 {} and num2 {} is {} ".format(num1,num2,num1+num2)) 
elif operation=='subtraction':
    print("we are doing subtraction operation is")
    print("the subtraction of num1 {} and num2 {} is {} ".format(num1,num2,num1-num2))
elif operation=='multipulication':
    print("we are doing multipulication")
    print("the multipulication of num1 {} and num2 {} is {} ".format(num1,num2,num1*num2))
elif operation=='division':
    print("we are doing division")
    print("the division of num1 {} and num2 {} is {} ".format(num1,num2,num1/num2))
    
else:
    print("provide suitable operation")
```
