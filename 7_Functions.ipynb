##### Local variables

### local variable:
      - the variable provided inside the function.
      - local variable values can not use outside the function.
      - until unless you return those values

```python
def addition():
    n1=10
    n2=20
    add=n1+n2
    print("the addition of {} and {} is {} ".format(n1,n2,add))
addition()
```
