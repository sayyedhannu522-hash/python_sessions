```python
# wap ask the user get a 10 random numbers between 0 to 10
# two count how many are greater than 5 
# how many are equal=0


import random
i=0
count=0 
count1=0
while i<10:
    ran_numm=random.randint(0,10)
    if ran_numm>5:
        print(ran_numm,end=' ')
        count=count+1 
    if ran_numm==0:       
        print(ran_numm,end=' ')
        count1=count1+1
    i=i+1    
print("/")
print(count)
print("/")
print(count1)
```
